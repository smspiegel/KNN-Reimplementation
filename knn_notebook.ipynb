## Algorithm Class Definition

```python
class KNN:
    def __init__(self,k,dist_metric,**kwargs):
        self.k = k
        self.dist_metric = dist_metric
        self.p = kwargs.get("p", None) # optional of using non euclid / manhattan distances

        self.X_memory = None
        self.y_memory = None
        #self.y_unencode_dict = None
        self.y_unencode_list = None
        self.y_unique = None # for unencoding

    def encode(self,y):
        #self.y_unique = set(y)
        y = y.squeeze(axis=1) # Transforms dataframe to pandas series, preserves order

        #self.y_unencode_dict = {}
        self.y_unencode_list = []

        #if self.y_unencode_dict is None: # Not previously encoded data

        for i,y_class in enumerate(list(y.unique())):
            self.y_unencode_list.append(y_class) # i is implied as the index anyways...
            #self.y_unencode_dict[i] = y_class

        self.y_memory = [] # may need to make an np.array([]) with dim determined by X array dims.
        for each_predictor in y:
            self.y_memory.append(self.y_unencode_list.index(each_predictor))
            #self.y_memory.append(self.y_unencode_dict[each_predictor])
        #return self.y_memory # no return statement needed
        return None


    def unencode(self,list_of_K_Neighbours): #Take in a single vector of encoded class values, spit out a single vector of classes
        '''This function specifically needs to be refactored and then implemented into the .predict route'''
        list_of_classes = []
        for encoded_class in list_of_K_Neighbours:
            list_of_classes.append(self.y_unencode_list[int(encoded_class)])
        print(list_of_classes)
        return list_of_classes

    def minkowski_dist(self,x,X_val): # Does not implement Hamming dist currently
        #p = kwargs.get("p", None) # already handled in init
        if self.dist_metric == "euclidean":
            self.p = 1
        if self.dist_metric == "manhattan":
            self.p = 2
        n = x.shape[0]
        print(n)

        sum = 0
        for i in range(n):
            sum += (np.abs(x[i] - X_val[i]))**self.p
        return sum**(1/self.p)


    def fit(self,X,y):
        # may need to encode y first if not all values ints 
        #if type(y.iat[0,0]) != int: need_to_encode=True
        #else: need_to_encode=False
        y = self.encode(y)#,need_to_encode)

        self.X_memory = X.to_numpy()
        #self.X_memory = [X[i] for i in range(len(X))]    # self. saves the data to this object, 
                                                    # otherwise values are lost after method finishes
   
    def predict(self,x):
        x = x.to_numpy()
        print(x[1,:])
        print(x)
        print(x.shape[0])
        First_K_values = []
        for j in range(x.shape[0]) :#for j samples in x

            #distances = list(map(lambda X_value: self.minkowski_dist(x=np.array(x[j,:]),X_val=X_value,p=self.p), self.X_memory))
            # Can also just use list comprehension I suppose
            print("STUFF",[x_value for x_value in self.X_memory])
            print("SHAPE OF X",np.array(x[j,:]).shape)
            print("THE ARRAY",x[j,:])
            distances = [self.minkowski_dist(x=np.array(x[j,:]),X_val=x_value) for x_value in self.X_memory]

            X = np.array(self.X_memory)
            y = np.array(self.y_memory)
            d = np.array(distances)
            dXy = None
            dXy = np.column_stack((d,X,y))
            sorted_distances = None
            sorted_distances = dXy[dXy[:,0].argsort()] # Sorts by first column vals (distances)

            first_k = sorted_distances[:self.k, -1] # Gives classes of first k sorted values
            First_K_values.append(first_k) # These need to have voting applied later...

        return First_K_values
        
```

## Iris Dataset

### Dataset Import

```python
data = pd.read_csv('datasets/iris/iris.csv')#,delimiter=",")
#print(data)
train = data[:-10]
test = data[-10:]

train_x = train.iloc[:,[0,1,2,3]].copy()
train_y = train.iloc[:,[-1]].copy()

test_x = test.iloc[:,[0,1,2,3]].copy()
test_y = test.iloc[:,[-1]].copy()
#print(train_y)
```

### Fitting to Data

```python
knn_iris = KNN(3,"euclidean")
```

```python
knn_iris.fit(train_x,train_y)
```

```python
print(knn_iris.y_unique)
```

### Results

```python
k_neighbour_vectors = knn_iris.predict(test_x)
k_neighbour_vectors
```

```python
for each_vector in k_neighbour_vectors:
    knn_iris.unencode(each_vector)
```
