## Algorithm Class Definition

```python
class KNN:
    def __init__(self,k,dist_metric,**kwargs):
        self.k = k
        self.dist_metric = dist_metric
        self.p = kwargs.get("p", None) # optional of using non euclid / manhattan distances

        self.X_memory = None
        self.y_memory = None
        self.y_unique = None # for unencoding

    def encode(self,y):
        #self.y_unique = set(y)
        if self.y_unique is None: # Not previously encoded data
            self.y_unique = list(set(y)) # to retain order            
        #else: self.y_unique = y_unique # unecessary line

        self.y_memory = [] # may need to make an np.array([]) with dim determined by X array dims.
        for each_predictor in y:
            self.y_memory.append(self.y_unique.index(each_predictor))
        #return self.y_memory # no return statement needed

    def minkowski_dist(x,X_val,dist_metric,p): # Does not implement Hamming dist currently
        #p = kwargs.get("p", None) # already handled in init
        if dist_metric == "euler":
            p = 1
        if dist_metric == "manhattan":
            p = 2
        n = len(x)

        sum = 0
        for i in range(n):
            sum += (np.abs(x - X_val))**p
        return sum**(1/p)


    def fit(self,X,y):
        if type(y[0]) != int: # may need to encode y first if not all values ints 
            y = encode(y)

        self.X_memory = [X[i] for i in range(len(X))]    # self. saves the data to this object, 
                                                    # otherwise values are lost after method finishes
   
    def predict(self,x):
        distances = list(map(lambda X_value: minkowski_dist(x=x,X_val=X_value,p=self.p), self.X_memory))
        # Can also just use list comprehension I suppose
        # distances = [minkowski_dist(x=x,X_val=x_value,p=p) for x_value in self.X_memory]
        # Keep a dictionary of original indices, and sorted index
        #Dict = {}

        X = np.array(self.X_memory)
        y = np.array(self.y_memory)
        d = np.array(distances)
        dXy = np.column_stack(d,X,y)
        sorted_distances = dXy[dXy[:,0].argsort()] # Sorts by first column vals (distances)

        first_k = sorted_distances[:self.k, -1] # Gives classes of first k sorted values

        return first_k
        
```
